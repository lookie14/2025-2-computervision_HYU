```python
img = Image.open("lena_color.png")
img = np.array(img)

pad_size = 2
(width, height, channel) = img.shape

padding_image = np.zeros((width+pad_size*2, height+pad_size*2, channel))

padding_image[pad_size:width+pad_size,pad_size:height+pad_size,:] = img

padding_image.shape
```

```python
img = Image.fromarray(img)
```

```python
#Gaussian filter(kernel)

def create_gaussian(size, sigma = 1):
    ax = np.arange(-size//2+1, size//2+1)
    xx, yy = np.meshgrid(ax, ax)

    kernel = np.exp(-(xx**2 + yy**2) / (2. * sigma**2))

    return kernel / np.sum(kernel)
```

```python
G_filter = create_gaussian(5, 1)
G_filter.shape
```

```python
padding_image = np.array(padding_image)
filtered_img = np.zeros((width, height, channel))

filter_size = 5
sigma = 1

G_filter = create_gaussian(filter_size, sigma)

for x in range(filter_size//2, width + filter_size//2):
    for y in range(filter_size//2, height + filter_size//2):
        for c in range(channel):
            for i in range(filter_size):
                for j in range(filter_size):
                    filtered_img[x][y][c] += padding_image[x+(i-filter_size//2)][y+(j-filter_size//2)][c]*G_filter[i][j]

filtered_img = Image.fromarray(filtered_img)


filtered_img
```
