```python
import numpy as np
from PIL import Image

img = Image.open("D:/coding/CV/week2/7577.png")
img
```

```python
print(img.size)

(width, height) = img.size
```

```python
for j in range(width):
    for i in range(height):
        print((np.asarray(img)[j][i]), '\t', end = '')
    print()
```

```python
img = np.array(img)

for x in range(width):
    for y in range(height):
        if(x>3 and x<24 and y > 3 and y < 24):
            img[x][y] = 100
        else:
            img[x][y] = 0

img = Image.fromarray(img)
img
```

```python
img2 = Image.open("D:/coding/CV/week2/lena_color.png")
img2
```

```python
print(img2.size)
img2 = np.array(img2)
print(img2.shape)
```

```python
(height, width, channel) = img2.shape

for x in range(width):
    for y in range(height):
        for c in range(channel):
            if c ==1 or c==2:
                img2[x][y][c] = 0

img2 = Image.fromarray(img2)
img2
```

```python
img2 = img2.rotate(90)
img2
```

```python
img2 = img2.transpose(Image.FLIP_TOP_BOTTOM)
img2
```

```python
img2 = img2.convert("L")
img2
```

```python
img2 = img2.convert('RGB')
img2
```

```python
img2 = np.array(img2)

print(img2.shape)
```

```python
(w, h) = img2.shape

for x in range(w):
    for y in range(h):
        if(img2[x][y] > 120) :
            img2[x][y] = 255
        else: img2[x][y] = 0

img2 = Image.fromarray(img2)

img2
```
